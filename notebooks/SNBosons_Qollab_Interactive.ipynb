```python
import numpy as np
from IPython.display import Audio, display

from qsong.harmony import TWINKLE, ODE_TO_JOY
from qsong.encode import events_from_tune, roundtrip_melody
from qsong.measure import measure_chsh_statistic, theoretical_chsh_s
from qsong.cli import run_experiment
from qsong.ionq_adapter import generate_qasm_bell_chsh

print("SNBosons core library loaded successfully!")
```

## 1. 11-Qubit Quantum Basis Round-Trip Memory
Monophonic melody events are packed into 11-qubit orthogonal computational basis states: bits 0-6 store MIDI pitch, bits 7-10 store 16th duration.

```python
events = events_from_tune(TWINKLE)
reconstructed = roundtrip_melody(events)

all_exact = all(orig == rec for orig, rec in zip(events, reconstructed))
print(f"Round trip tested on {len(events)} notes.")
print(f"Deterministic fidelity P=1.0 exact recovery: {all_exact}")
```

## 2. CHSH Bell Inequality Verification
Measurement angles derived from melodic direction and pitch register violate the classical Bell-CHSH ceiling ($S \le 2.0$).

```python
s_theory = theoretical_chsh_s()
s_empirical, correlations = measure_chsh_statistic(shots_per_setting=20000, seed=2026)

print(f"Theoretical Max CHSH S (Tsirelson Bound): {s_theory:.4f}")
print(f"Measured CHSH S (20,000 shots):           {s_empirical:.4f}")
print("Correlations per setting:", correlations)
assert s_empirical > 2.0, "Non-classical entanglement violation!"
```

## 3. Generate Quantum Duet vs Classical Control Audio

```python
# Execute quantum duet and independent classical baseline on Twinkle
res_quantum = run_experiment(tune=TWINKLE, seed=101, mode="quantum")
res_classical = run_experiment(tune=TWINKLE, seed=101, mode="independent")

print("=== OBJECTIVE METRICS ===")
print(f"Quantum CHSH S:               {res_quantum['chsh']:.4f}")
print(f"Classical CHSH S:             {res_classical['chsh']:.4f}")
print(f"Quantum Same Direction Motion: {res_quantum['same_direction_fraction']*100:.2f}%")
print(f"Classical Same Dir Motion:    {res_classical['same_direction_fraction']*100:.2f}%")
print(f"RMS Loudness Quantum:         {res_quantum['rms']:.4f}")
print(f"RMS Loudness Classical:       {res_classical['rms']:.4f}")
```

### Inline Audio Playback

```python
print("1. Quantum Entangled Duet:")
display(Audio(res_quantum["audio"], rate=44100))

print("2. Independent Classical Control:")
display(Audio(res_classical["audio"], rate=44100))
```
